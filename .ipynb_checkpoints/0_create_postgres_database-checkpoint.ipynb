```python
engine = create_engine(f"postgresql://{DB_USER}:{DB_PASSWORD}@{DB_HOST}:{DB_PORT}/{DB_NAME}")
```

```python
with engine.begin() as conn:

    conn.execute(text("""
            -- ---------------------------------------------------------
            -- dim_countries
            -- ---------------------------------------------------------
                CREATE TABLE dim_countries (
                country_id CHAR(2) PRIMARY KEY,
                country_name VARCHAR(100) NOT NULL,
                region VARCHAR(100),
                country_risk_base NUMERIC(4,3) CHECK (country_risk_base BETWEEN 0 AND 1),
                sanctions_risk NUMERIC(4,3) CHECK (sanctions_risk BETWEEN 0 AND 1),
                port_efficiency_base NUMERIC(4,3) CHECK (port_efficiency_base BETWEEN 0 AND 1),
                avg_customs_days INT,
                currency_code CHAR(3),
                language VARCHAR(100),
                wto_member BOOLEAN
            );

            -- ---------------------------------------------------------
            -- dim_product_categories
            -- ---------------------------------------------------------
            CREATE TABLE dim_product_categories (
                category_id INT PRIMARY KEY,
                category_name VARCHAR(150) NOT NULL,
                category_description TEXT
            );

            -- ---------------------------------------------------------
            -- dim_products
            -- ---------------------------------------------------------
            CREATE TABLE dim_products (
                product_id INT PRIMARY KEY,
                product_name VARCHAR(255) NOT NULL,

                category_id INT NOT NULL,
                unit VARCHAR(50),

                base_price_usd NUMERIC(14,2),
                weight_kg_per_unit NUMERIC(14,4),

                hs_code CHAR(6),

                perishable BOOLEAN,

                CONSTRAINT fk_products_category
                    FOREIGN KEY (category_id)
                    REFERENCES dim_product_categories(category_id)
            );

            -- ---------------------------------------------------------
            -- dim_transport_modes
            -- ---------------------------------------------------------
            CREATE TABLE dim_transport_modes (
                transport_mode_id INT PRIMARY KEY,
                mode_name VARCHAR(100) NOT NULL,

                avg_transit_days INT,

                cost_multiplier NUMERIC(8,3),

                delay_probability NUMERIC(4,3)
                    CHECK (delay_probability BETWEEN 0 AND 1),

                uses_port BOOLEAN
            );

            -- ---------------------------------------------------------
            -- dim_ports
            -- ---------------------------------------------------------
            CREATE TABLE dim_ports (
                port_id CHAR(5) PRIMARY KEY,
                port_name VARCHAR(255) NOT NULL,

                country_id CHAR(2) NOT NULL,

                efficiency_base NUMERIC(4,3)
                    CHECK (efficiency_base BETWEEN 0 AND 1),

                annual_teu_capacity BIGINT,

                CONSTRAINT fk_ports_country
                    FOREIGN KEY (country_id)
                    REFERENCES dim_countries(country_id)
            );

            -- ---------------------------------------------------------
            -- dim_suppliers
            -- ---------------------------------------------------------
            CREATE TABLE dim_suppliers (
                supplier_id INT PRIMARY KEY,

                supplier_name VARCHAR(255) NOT NULL,

                country_id CHAR(2) NOT NULL,

                city VARCHAR(150),

                primary_category_id INT,

                registration_date DATE,

                years_of_relationship NUMERIC(6,2),

                reliability_score NUMERIC(4,3)
                    CHECK (reliability_score BETWEEN 0 AND 1),

                maturity_score INT
                    CHECK (maturity_score BETWEEN 1 AND 20),

                payment_terms_days INT,

                min_order_usd NUMERIC(14,2),

                certifications JSONB,

                is_fraudster BOOLEAN DEFAULT FALSE,

                active BOOLEAN DEFAULT TRUE,

                CONSTRAINT fk_suppliers_country
                    FOREIGN KEY (country_id)
                    REFERENCES dim_countries(country_id),

                CONSTRAINT fk_suppliers_category
                    FOREIGN KEY (primary_category_id)
                    REFERENCES dim_product_categories(category_id)
            );

            -- ---------------------------------------------------------
            -- dim_supplier_products
            -- ---------------------------------------------------------
            CREATE TABLE dim_supplier_products (
                sp_id INT PRIMARY KEY,

                supplier_id INT NOT NULL,

                product_id INT NOT NULL,

                supplier_price_usd NUMERIC(14,2),

                lead_time_days INT,

                moq INT,

                CONSTRAINT fk_sp_supplier
                    FOREIGN KEY (supplier_id)
                    REFERENCES dim_suppliers(supplier_id),

                CONSTRAINT fk_sp_product
                    FOREIGN KEY (product_id)
                    REFERENCES dim_products(product_id),

                CONSTRAINT uq_supplier_product
                    UNIQUE (supplier_id, product_id)
            );

            -- ---------------------------------------------------------
            -- dim_customers
            -- ---------------------------------------------------------
            CREATE TABLE dim_customers (
                customer_id INT PRIMARY KEY,

                customer_name VARCHAR(255) NOT NULL,

                city VARCHAR(150),

                latitude NUMERIC(10,6),
                longitude NUMERIC(10,6),

                industry VARCHAR(150),

                segment VARCHAR(50),

                annual_revenue_mln_rub BIGINT,

                credit_limit_usd NUMERIC(16,2),

                inn CHAR(10) UNIQUE
            );

            -- ---------------------------------------------------------
            -- dim_fx_rates
            -- ---------------------------------------------------------
            CREATE TABLE dim_fx_rates (
                fx_id INT PRIMARY KEY,

                date DATE NOT NULL,

                usdrub NUMERIC(14,6),

                usdcny NUMERIC(14,6),

                usdeur NUMERIC(14,6),

                UNIQUE(date)
            );

            -- ---------------------------------------------------------
            -- dim_country_risk
            -- ---------------------------------------------------------
            CREATE TABLE dim_country_risk (
                cr_id INT PRIMARY KEY,

                country_id CHAR(2) NOT NULL,

                quarter_start DATE NOT NULL,

                country_risk_score NUMERIC(4,3)
                    CHECK (country_risk_score BETWEEN 0 AND 1),

                sanctions_risk_score NUMERIC(4,3)
                    CHECK (sanctions_risk_score BETWEEN 0 AND 1),

                political_stability NUMERIC(4,3)
                    CHECK (political_stability BETWEEN 0 AND 1),

                CONSTRAINT fk_country_risk_country
                    FOREIGN KEY (country_id)
                    REFERENCES dim_countries(country_id),

                CONSTRAINT uq_country_quarter
                    UNIQUE(country_id, quarter_start)
            );

            -- =========================================================
            -- FACT TABLES
            -- =========================================================

            -- ---------------------------------------------------------
            -- fact_purchase_orders
            -- ---------------------------------------------------------
            CREATE TABLE fact_purchase_orders (
                po_id INT PRIMARY KEY,

                order_date DATE NOT NULL,

                supplier_id INT NOT NULL,

                product_id INT NOT NULL,

                transport_mode_id INT NOT NULL,

                origin_port_id CHAR(5),

                destination_port_id CHAR(5),

                quantity INT,

                unit_price_usd NUMERIC(14,2),

                total_value_usd NUMERIC(16,2),

                currency_code CHAR(3),

                fx_rate_to_usd NUMERIC(16,6),

                total_value_local NUMERIC(16,2),

                freight_cost_usd NUMERIC(14,2),

                insurance_cost_usd NUMERIC(14,2),

                customs_duty_usd NUMERIC(14,2),

                incoterm VARCHAR(20),

                payment_method VARCHAR(100),

                expected_delivery_days INT,

                actual_delivery_days INT,

                port_congestion_at_order NUMERIC(4,3)
                    CHECK (port_congestion_at_order BETWEEN 0 AND 1),

                shock_index_at_order NUMERIC(4,3)
                    CHECK (shock_index_at_order BETWEEN 0 AND 1),

                supplier_reliability NUMERIC(4,3)
                    CHECK (supplier_reliability BETWEEN 0 AND 1),

                final_state VARCHAR(50),

                is_fraud BOOLEAN DEFAULT FALSE,

                fraud_type VARCHAR(50),

                CONSTRAINT fk_po_supplier
                    FOREIGN KEY (supplier_id)
                    REFERENCES dim_suppliers(supplier_id),

                CONSTRAINT fk_po_product
                    FOREIGN KEY (product_id)
                    REFERENCES dim_products(product_id),

                CONSTRAINT fk_po_transport
                    FOREIGN KEY (transport_mode_id)
                    REFERENCES dim_transport_modes(transport_mode_id),

                CONSTRAINT fk_po_origin_port
                    FOREIGN KEY (origin_port_id)
                    REFERENCES dim_ports(port_id),

                CONSTRAINT fk_po_destination_port
                    FOREIGN KEY (destination_port_id)
                    REFERENCES dim_ports(port_id)
            );

            -- ---------------------------------------------------------
            -- fact_po_state_log
            -- ---------------------------------------------------------
            CREATE TABLE fact_po_state_log (
                state_event_id INT PRIMARY KEY,

                po_id INT NOT NULL,

                state VARCHAR(100) NOT NULL,

                state_date DATE NOT NULL,

                duration_days INT,

                CONSTRAINT fk_state_log_po
                    FOREIGN KEY (po_id)
                    REFERENCES fact_purchase_orders(po_id)
            );

            -- ---------------------------------------------------------
            -- fact_sales_orders
            -- ---------------------------------------------------------
            CREATE TABLE fact_sales_orders (
                so_id INT PRIMARY KEY,

                sale_date VARCHAR(50),

                po_id INT,

                customer_id INT NOT NULL,

                product_id INT NOT NULL,

                quantity INT,

                unit_price_usd NUMERIC(14,2),

                discount_pct NUMERIC(5,2),

                total_value_usd NUMERIC(16,2),

                total_value_rub NUMERIC(16,2),

                fx_rate_usd_rub NUMERIC(14,6),

                payment_terms_days INT,

                delivery_city VARCHAR(150),

                CONSTRAINT fk_sales_po
                    FOREIGN KEY (po_id)
                    REFERENCES fact_purchase_orders(po_id),

                CONSTRAINT fk_sales_customer
                    FOREIGN KEY (customer_id)
                    REFERENCES dim_customers(customer_id),

                CONSTRAINT fk_sales_product
                    FOREIGN KEY (product_id)
                    REFERENCES dim_products(product_id)
            );

            -- ---------------------------------------------------------
            -- fact_supplier_audits
            -- ---------------------------------------------------------
            CREATE TABLE fact_supplier_audits (
                audit_id INT PRIMARY KEY,

                supplier_id INT NOT NULL,

                audit_date DATE NOT NULL,

                audit_type VARCHAR(100),

                auditor VARCHAR(255),

                findings_count INT,

                audit_score NUMERIC(4,3)
                    CHECK (audit_score BETWEEN 0 AND 1),

                passed BOOLEAN,

                notes TEXT,

                CONSTRAINT fk_audit_supplier
                    FOREIGN KEY (supplier_id)
                    REFERENCES dim_suppliers(supplier_id)
            );

            -- ---------------------------------------------------------
            -- fact_port_congestion
            -- ---------------------------------------------------------
            CREATE TABLE fact_port_congestion (
                pc_id INT PRIMARY KEY,

                date DATE NOT NULL,

                port_id CHAR(5) NOT NULL,

                congestion_index NUMERIC(4,3)
                    CHECK (congestion_index BETWEEN 0 AND 1),

                vessels_waiting INT,

                CONSTRAINT fk_congestion_port
                    FOREIGN KEY (port_id)
                    REFERENCES dim_ports(port_id)
            );

            -- =========================================================
            -- INDEXES
            -- =========================================================

            CREATE INDEX idx_products_category
                ON dim_products(category_id);

            CREATE INDEX idx_suppliers_country
                ON dim_suppliers(country_id);

            CREATE INDEX idx_supplier_products_supplier
                ON dim_supplier_products(supplier_id);

            CREATE INDEX idx_supplier_products_product
                ON dim_supplier_products(product_id);

            CREATE INDEX idx_po_supplier
                ON fact_purchase_orders(supplier_id);

            CREATE INDEX idx_po_product
                ON fact_purchase_orders(product_id);

            CREATE INDEX idx_po_order_date
                ON fact_purchase_orders(order_date);

            CREATE INDEX idx_sales_customer
                ON fact_sales_orders(customer_id);

            CREATE INDEX idx_sales_product
                ON fact_sales_orders(product_id);

            CREATE INDEX idx_port_congestion_date
                ON fact_port_congestion(date);

            CREATE INDEX idx_country_risk_country
                ON dim_country_risk(country_id);

    """))
```

```python
# =========================================================
# DATA CLEANING
# =========================================================

def clean_dataframe(df: pd.DataFrame) -> pd.DataFrame:
    """
    Basic dataframe cleaning before PostgreSQL insert.
    """

    # Replace NaN with None for PostgreSQL
    df = df.replace({np.nan: None})

    return df


# =========================================================
# DATE CONVERSIONS
# =========================================================

date_columns = {
    "df_suppliers": ["registration_date"],
    "df_fx": ["date"],
    "df_country_risk": ["quarter_start"],
    "df_po": ["order_date"],
    "df_po_states": ["state_date"],
    "df_audits": ["audit_date"],
    "df_port_congestion": ["date"],
}

for df_name, cols in date_columns.items():

    df_obj = globals()[df_name]

    for col in cols:
        if col in df_obj.columns:
            df_obj[col] = pd.to_datetime(
                df_obj[col],
                errors="coerce"
            ).dt.date

# =========================================================
# CLEAN ALL DATAFRAMES
# =========================================================

dfs = [
    df_countries,
    df_categories,
    df_products,
    df_transport_modes,
    df_ports,
    df_suppliers,
    df_supplier_products,
    df_customers,
    df_fx,
    df_country_risk,
    df_po,
    df_po_states,
    df_so,
    df_audits,
    df_port_congestion
]

dfs = [clean_dataframe(df) for df in dfs]

(
    df_countries,
    df_categories,
    df_products,
    df_transport_modes,
    df_ports,
    df_suppliers,
    df_supplier_products,
    df_customers,
    df_fx,
    df_country_risk,
    df_po,
    df_po_states,
    df_so,
    df_audits,
    df_port_congestion
) = dfs

# =========================================================
# LOAD TABLES IN FK ORDER
# =========================================================

load_order = [

    # DIMENSIONS WITHOUT FKs
    #("dim_countries", df_countries),
    #("dim_product_categories", df_categories),
    #("dim_transport_modes", df_transport_modes),
    #("dim_fx_rates", df_fx),

    # DEPENDENT DIMENSIONS
    #("dim_products", df_products),
    #("dim_ports", df_ports),
    #("dim_customers", df_customers),

    #("dim_suppliers", df_suppliers),

    #("dim_supplier_products", df_supplier_products),

    #("dim_country_risk", df_country_risk),

    # FACT TABLES
    #("fact_purchase_orders", df_po),

    #("fact_po_state_log", df_po_states),

    ("fact_sales_orders", df_so),

    ("fact_supplier_audits", df_audits),

    ("fact_port_congestion", df_port_congestion),
]

# =========================================================
# INSERT DATA
# =========================================================

for table_name, df in load_order:

    print(f"Loading {table_name}...")

    dtype_mapping = {}

    # JSONB handling
    # if table_name == "dim_suppliers":
    #     dtype_mapping["certifications"] = JSON

    df.to_sql(
        name=table_name,
        con=engine,
        schema="public",
        if_exists="append",
        index=False,
        chunksize=5000,
        method="multi",
        dtype=dtype_mapping
    )

    print(f"{table_name}: {len(df):,} rows inserted")

print("\nAll tables loaded successfully.")
```
